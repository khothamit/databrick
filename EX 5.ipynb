```sql
%sql
USE CATALOG adventure_works;
USE SCHEMA bronze;

SHOW TABLES;
```

```sql
%sql
SELECT *
FROM sales_order_detail
ORDER BY ModifiedDate DESC
LIMIT 10;
```

```sql
%sql
SELECT
    COUNT(*) AS total_sales_rows,
    COUNT(DISTINCT ProductID) AS distinct_products_sold
FROM sales_order_detail;
```

```sql
%sql
SELECT
    COUNT(DISTINCT ProductID) AS distinct_products_in_catalog
FROM product;
```

```sql
%sql
SELECT
    f.ProductID,
    p.Name AS product_name,
    SUM(f.LineTotal) AS total_revenue
FROM sales_order_detail AS f
LEFT JOIN product AS p
    ON f.ProductID = p.ProductID
GROUP BY
    f.ProductID,
    p.Name
ORDER BY total_revenue DESC
LIMIT 10;
```

```sql
%sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(CASE WHEN LineTotal IS NULL THEN 1 END) AS null_line_totals,
    COUNT(CASE WHEN LineTotal < 0 THEN 1 END) AS negative_line_totals,
    COUNT(
        CASE WHEN LineTotal IS NULL OR LineTotal < 0 THEN 1 END
    ) AS total_flagged_rows
FROM sales_order_detail;
```

```sql
%sql
SELECT *
FROM sales_order_detail
WHERE LineTotal IS NULL
   OR LineTotal < 0;
```
