```sql
SHOW TABLES IN adventure_works.bronze;
```

```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT SalesOrderDetailID) AS unique_sales_line_ids,
    COUNT(*) - COUNT(SalesOrderDetailID) AS missing_ids
FROM adventure_works.bronze.sales_order_detail;
```

```sql
SELECT 'fact_sales' AS table_name, COUNT(*) AS row_count
FROM adventure_works.bronze.sales_order_detail

UNION ALL

SELECT 'dim_product', COUNT(*)
FROM adventure_works.bronze.product

UNION ALL

SELECT 'dim_sales_territory', COUNT(*)
FROM adventure_works.bronze.sales_territory;
```

```sql
SELECT 'before_join' AS check_name, COUNT(*) AS row_count
FROM adventure_works.bronze.sales_order_detail

UNION ALL

SELECT 'after_product_left_join', COUNT(*)
FROM adventure_works.bronze.sales_order_detail AS f
LEFT JOIN adventure_works.bronze.product AS p
    ON f.ProductID = p.ProductID;
```
